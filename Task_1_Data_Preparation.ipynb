### Objective
The objective of this task is to load and explore the telecommunications customer churn dataset, identify and handle missing values, perform necessary data preprocessing, encode categorical variables into numerical representations, and split the dataset into training and testing sets for further analysis and model evaluation.

### Dataset
Telco Customer Churn Dataset

```

# Step 2: Import Required Libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Display configuration
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("All required libraries imported successfully!")
```

```

# Step 3: Load Dataset

from google.colab import files

uploaded = files.upload()

file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
```

```

# Step 4: Display First Five Records

df.head()
```

```

# Step 5: Dataset Dimensions

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
print("Dataset Shape:", df.shape)
```

```

# Step 6: Dataset Information

df.info()
```

```

# Step 7: Check Missing Values

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```

# Step 8: Convert TotalCharges into Numeric Format

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

print("Updated Data Type:")
print(df["TotalCharges"].dtype)

print("\nMissing TotalCharges Values:")
print(df["TotalCharges"].isnull().sum())
```

```

# Step 9: Investigate Missing Records

missing_records = df[df["TotalCharges"].isnull()]

print("Records with Missing TotalCharges:")
display(missing_records)

print("\nTenure of Missing Records:")
print(missing_records["tenure"].value_counts())

print("\nChurn Status:")
print(missing_records["Churn"].value_counts())
```

```

# Step 10: Handle Missing Values

# Verify that all missing TotalCharges records have zero tenure
assert missing_records["tenure"].eq(0).all(), \
    "Unexpected missing records found. Investigate before filling."

# Replace missing TotalCharges with 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)

print("Missing values handled successfully!")

print("\nRemaining Missing Values:")
print(df.isnull().sum().sum())
```

```

# Step 11: Check Duplicate Records

duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)

# Remove duplicates if present
df = df.drop_duplicates()

print("Dataset Shape After Duplicate Removal:", df.shape)
```

```

# Step 12: Identify Categorical and Numerical Columns

categorical_columns = df.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("Categorical Columns:")
print(categorical_columns)

print("\nNumerical Columns:")
print(numerical_columns)
```

```

# Step 13: Remove Customer ID from Features

df_clean = df.copy()

X = df_clean.drop(
    columns=["customerID", "Churn"]
)

y = df_clean["Churn"].map({
    "No": 0,
    "Yes": 1
})

print("Feature Dataset Shape:", X.shape)
print("Target Dataset Shape:", y.shape)

print("\nTarget Variable Distribution:")
print(y.value_counts())
```

```

# Step 14: Train-Test Split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

```

# Step 15: Categorical Variable Encoding

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Identify categorical and numerical features
categorical_features = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Create OneHotEncoder with compatibility for older versions
try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

# Create preprocessing transformer
preprocessor = ColumnTransformer(
    transformers=[
        ("numerical", "passthrough", numerical_features),
        ("categorical", encoder, categorical_features)
    ]
)

# Fit on training data and transform both datasets
X_train_encoded_array = preprocessor.fit_transform(X_train)

X_test_encoded_array = preprocessor.transform(X_test)

print("Categorical Encoding Completed!")
```

```

# Step 16: Create Encoded DataFrames

feature_names = preprocessor.get_feature_names_out()

X_train_encoded = pd.DataFrame(
    X_train_encoded_array,
    columns=feature_names,
    index=X_train.index
)

X_test_encoded = pd.DataFrame(
    X_test_encoded_array,
    columns=feature_names,
    index=X_test.index
)

print("Encoded Training Dataset:")
display(X_train_encoded.head())

print("\nEncoded Testing Dataset:")
display(X_test_encoded.head())
```

```

# Step 17: Final Dataset Verification

print("========== FINAL DATASET SUMMARY ==========")

print("\nOriginal Dataset Shape:", df.shape)

print("Training Dataset Shape:", X_train_encoded.shape)

print("Testing Dataset Shape:", X_test_encoded.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())

print("\nMissing Values in Training Data:",
      X_train_encoded.isnull().sum().sum())

print("Missing Values in Testing Data:",
      X_test_encoded.isnull().sum().sum())

print("\nAll preprocessing verification completed!")
```

```

# Step 18: Save Processed Datasets

import os
import shutil

output_folder = "Task_1_Outputs"

os.makedirs(output_folder, exist_ok=True)

# Save cleaned original dataset
df.to_csv(
    f"{output_folder}/cleaned_telco_dataset.csv",
    index=False
)

# Save encoded training and testing features
X_train_encoded.to_csv(
    f"{output_folder}/X_train.csv",
    index=False
)

X_test_encoded.to_csv(
    f"{output_folder}/X_test.csv",
    index=False
)

# Save target variables
y_train.to_csv(
    f"{output_folder}/y_train.csv",
    index=False
)

y_test.to_csv(
    f"{output_folder}/y_test.csv",
    index=False
)

# Create ZIP archive
shutil.make_archive(
    "Task_1_Outputs",
    "zip",
    output_folder
)

print("All Task 1 output files saved successfully!")
```
